Hour Ending appears to be local time. 8 regional load columns and one total ERCOT column. LEt's confirm the columns add up.

```python
regional_total=df.iloc[:,1:9].sum(axis=1)
ercot_total=df.iloc[:,9]
df.loc[~np.isclose(regional_total,ercot_total)]
reg_erc_diff=ercot_total-regional_total
(reg_erc_diff.abs() > 1).sum()
df["reg_sum"]=regional_total
df["reg_erc_diff"]=reg_erc_diff
df["perc_diff"]=100*(reg_erc_diff/ercot_total)
df.loc[reg_erc_diff.abs() > 1, ["ERCOT","reg_sum","reg_erc_diff","perc_diff"]].describe()
list(enumerate(df.columns))
```
